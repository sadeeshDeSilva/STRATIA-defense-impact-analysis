```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
from pmdarima import auto_arima
from statsmodels.tsa.arima.model import ARIMA
```

```python

path = "/content/drive/MyDrive/Final_Master_Dataset.csv"

df = pd.read_csv(path)
df.head()
```

```python
df = df[["Country", "Year", "GDP_Growth_Annual"]].dropna()
df["Year"] = df["Year"].astype(int)

df.head()

# Function for train ARIMA for each country:

def forecast_last_5_years(country_name, forecast_years=5):
    data = df[df["Country"] == country_name].sort_values("Year")

    ts = data["GDP_Growth_Annual"].values
    years = data["Year"].values

# Last 5 years (test)
    test_years = years[-5:]
    test_values = ts[-5:]

# Training data (all years except last 5)
    train_values = ts[:-5]

    model = ARIMA(train_values, order=(2,1,2))
    model_fit = model.fit()

    # Predict the last 5 actual years for testing..
    pred_test = model_fit.forecast(steps=5)

# Metrics
    mae = mean_absolute_error(test_values, pred_test)
    rmse = np.sqrt(mean_squared_error(test_values, pred_test))
    r2 = r2_score(test_values, pred_test)

    # Forecast exactly 5 periods (same as test)
    predicted_values = model_fit.forecast(steps=5)


    return {
        "country": country_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Test_Years": list(test_years),
        "Actual_Last5": list(test_values),
        "Pred_Last5": list(predicted_values),

    }

countries = df["Country"].unique()

results = []

for c in countries:
    print(f"Running ARIMA for {c}...")
    res = forecast_last_5_years(c)
    results.append(res)
```

```python
for r in results:
    print("\nCountry:", r["country"])
    print("Test Years:      ", r["Test_Years"])
    print("Actual Values:   ", r["Actual_Last5"])
    print("Predicted Values:", r["Pred_Last5"])
    print("MAE :", round(r["MAE"], 3))
    print("RMSE:", round(r["RMSE"], 3))
    print("R²  :", round(r["R2"], 3))

```

```python

from statsmodels.tsa.statespace.sarimax import SARIMAX

# Keep only needed columns;
df = df[["Country", "Year", "GDP_Growth_Annual"]].dropna()
df["Year"] = df["Year"].astype(int)

# Function:
def forecast_last_5_years_sarima(country_name, forecast_years=5):

    data = df[df["Country"] == country_name].sort_values("Year")
    ts = data["GDP_Growth_Annual"].values
    years = data["Year"].values

    # Last 5 years (test)..
    test_years = years[-5:]
    test_values = ts[-5:]

    # Training data..
    train_values = ts[:-5]

    # SARIMA model:
    model = SARIMAX(
        train_values,
        order=(2, 1, 2),
        seasonal_order=(1, 0, 1, 4),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    model_fit = model.fit(disp=False)

    # Forecast next 5 (same as the test years)..
    pred_test = model_fit.forecast(steps=5)

    # Metrics:
    mae = mean_absolute_error(test_values, pred_test)
    rmse = np.sqrt(mean_squared_error(test_values, pred_test))
    r2 = r2_score(test_values, pred_test)

    return {
        "country": country_name,
        "Test_Years": list(test_years),
        "Actual_Last5": list(test_values),
        "Pred_Last5": list(pred_test),
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }

# Run SARIMA for all countries..
countries = df["Country"].unique()
results = []

for c in countries:
    print(f"Running SARIMA for {c}...")
    res = forecast_last_5_years_sarima(c)
    results.append(res)

# Print all results...
for r in results:
    print("\n-----------------------------------------")
    print("Country:", r["country"])
    print("Test Years:      ", r["Test_Years"])
    print("Actual Values:   ", r["Actual_Last5"])
    print("Predicted Values:", np.round(r["Pred_Last5"], 3))
    print("MAE :", round(r["MAE"], 3))
    print("RMSE:", round(r["RMSE"], 3))
    print("R²  :", round(r["R2"], 3))
```

```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline


INPUT = "/content/drive/MyDrive/Final_Master_Dataset.csv"

df = pd.read_csv(INPUT)

# Sort:
df = df.sort_values(["Country", "Year"])

# Create GDP growth target:
df["GDP_growth"] = (df["GDP_Growth_Annual"].shift(-1) - df["GDP_Growth_Annual"]) / df["GDP_Growth_Annual"]
df = df.groupby("Country").apply(lambda x: x.iloc[:-1]).reset_index(drop=True)

# Select features:
FEATURES = ["Country", "D_Expenditure_GDP", "GDP_Growth_Annual"]

df = df.dropna(subset=FEATURES + ["GDP_growth"])

X = df[FEATURES]
y = df["GDP_growth"]

# Preprocessing:
numeric_features = ["D_Expenditure_GDP", "GDP_Growth_Annual"]
categorical_features = ["Country"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(drop="first"), categorical_features)
    ]
)

# Model:
model = Pipeline([
    ("pre", preprocessor),
    ("xgb", XGBRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.9,
        colsample_bytree=0.8,
        random_state=42
    ))
])

# Train-test split by random:
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=True, random_state=42
)

# Train:
model.fit(X_train, y_train)

# Predict:
y_pred = model.predict(X_test)


results = X_test.copy()
results["Actual_GDP_Growth"] = y_test.values
results["Predicted_GDP_Growth"] = y_pred
results["Error"] = results["Actual_GDP_Growth"] - results["Predicted_GDP_Growth"]

results = results.reset_index(drop=True)

# Add Year column:
merged = results.merge(
    df[["Country", "Year", "GDP_growth"]],
    left_on=["Country", "Actual_GDP_Growth"],
    right_on=["Country", "GDP_growth"],
    how="left"
)

# Drop duplicate GDP_growth column:
merged = merged.drop(columns=["GDP_growth"])

# Show first 10 rows:
print("\nSAMPLE PREDICTIONS (First 10 rows)")
print(merged[["Country", "Year", "Actual_GDP_Growth", "Predicted_GDP_Growth", "Error"]].head(10))

# Metrics:
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(((y_test - y_pred)**2).mean())
r2 = r2_score(y_test, y_pred)

print("\nXGBoost Results")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```
